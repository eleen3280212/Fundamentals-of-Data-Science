```python
import pandas as pd

df = pd.read_csv("student-mat.csv", sep=";")
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
x=df.drop('G3',axis=1)
y=df["G3"]
```

```python
x = pd.get_dummies(x, drop_first=True)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)

print("R-square =", r2)
```

The R-square value is 0.724, which means that the linear regression model explains approximately 72.4% of the variation in the final grade (G3). This indicates that the model has a good fit and provides reasonably accurate predictions, while the remaining 27.6% of the variation may be explained by other factors not included in the model.

```python
importance = pd.DataFrame({
    "Feature": x.columns,
    "Coefficient": model.coef_})

importance["Absolute"] = abs(importance["Coefficient"])

importance = importance.sort_values(
    by="Absolute",
    ascending=False)

importance.head(10)
```

The most significant variable is G2, with the largest absolute coefficient (0.958), indicating that the second period grade is the strongest predictor of the final grade (G3). Other important variables include schoolsup_yes, reason_home, activities_yes, and Fjob_health, which also have relatively large coefficients and contribute to the prediction of student performance.
